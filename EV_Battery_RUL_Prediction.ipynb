```python
discharge_data = []

for cycle in cycles:
    cycle_type = cycle["type"][0]

    if cycle_type == "discharge":
        data = cycle["data"][0, 0]
        capacity = data["Capacity"]

        discharge_data.append(float(capacity))

print("Number of discharge cycles:", len(discharge_data))
print("First 5 capacities:", discharge_data[:5])
```

```python
import pandas as pd

battery_data = pd.DataFrame({
    "Cycle": range(1, len(discharge_data) + 1),
    "Capacity_Ah": discharge_data
})

battery_data.head()
```

```python
battery_data.describe()
```

```python
battery_data["SOH"] = (
    battery_data["Capacity_Ah"] / battery_data["Capacity_Ah"].iloc[0]
) * 100

battery_data.head()
```

```python
print("Initial SOH:", battery_data["SOH"].iloc[0])
print("Final SOH:", battery_data["SOH"].iloc[-1])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.plot(
    battery_data["Cycle"],
    battery_data["Capacity_Ah"]
)

plt.xlabel("Cycle")
plt.ylabel("Capacity (Ah)")
plt.title("Battery Capacity Degradation - B0005")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    battery_data["Cycle"],
    battery_data["SOH"]
)

plt.xlabel("Cycle")
plt.ylabel("State of Health (%)")
plt.title("Battery SOH Degradation - B0005")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(battery_data["Cycle"], battery_data["SOH"], label="SOH")

plt.axhline(
    80,
    linestyle="--",
    label="80% SOH"
)

plt.xlabel("Cycle")
plt.ylabel("State of Health (%)")
plt.title("Battery SOH Degradation - B0005")
plt.legend()

plt.show()
```

```python
discharge_features = []

for cycle in cycles:
    if cycle["type"][0] == "discharge":
        data = cycle["data"][0, 0]

        voltage = data["Voltage_measured"].flatten()
        current = data["Current_measured"].flatten()
        temperature = data["Temperature_measured"].flatten()

        discharge_features.append({
            "Voltage_mean": voltage.mean(),
            "Voltage_min": voltage.min(),
            "Voltage_max": voltage.max(),
            "Current_mean": current.mean(),
            "Temperature_mean": temperature.mean(),
            "Temperature_max": temperature.max()
        })

print("Number of discharge cycles:", len(discharge_features))
```

```python
features = pd.DataFrame(discharge_features)

features.head()
```

```python
battery_data = pd.concat(
    [battery_data, features],
    axis=1
)

battery_data.head()
```

```python
print(battery_data.columns)
```

```python
eol_capacity = 1.4

eol_cycle = battery_data[
    battery_data["Capacity_Ah"] <= eol_capacity
]["Cycle"].min()

print("EOL cycle:", eol_cycle)
```

```python
if pd.isna(eol_cycle):
    print("B0005 did not reach the 1.4 Ah EOL threshold.")
else:
    battery_data["RUL"] = eol_cycle - battery_data["Cycle"]
    print(battery_data[["Cycle", "Capacity_Ah", "SOH", "RUL"]].head())
```

```python
eol_cycle = battery_data[
    battery_data["Capacity_Ah"] <= 1.4
]["Cycle"].iloc[0]

print("EOL cycle:", eol_cycle)
```

```python
battery_data = battery_data[
    battery_data["Cycle"] <= eol_cycle
].copy()

battery_data["RUL"] = eol_cycle - battery_data["Cycle"]

print(battery_data.tail())
```

```python
print("Number of rows:", len(battery_data))
print("Minimum RUL:", battery_data["RUL"].min())
print("Maximum RUL:", battery_data["RUL"].max())
```

```python
battery_data.head()
```

```python
split = int(len(battery_data) * 0.8)

train_data = battery_data.iloc[:split].copy()
test_data = battery_data.iloc[split:].copy()

print("Training rows:", len(train_data))
print("Testing rows:", len(test_data))

print("Training cycles:", train_data["Cycle"].min(), "to", train_data["Cycle"].max())
print("Testing cycles:", test_data["Cycle"].min(), "to", test_data["Cycle"].max())
```

```python
features = [
    "Cycle",
    "Voltage_mean",
    "Voltage_min",
    "Voltage_max",
    "Current_mean",
    "Temperature_mean",
    "Temperature_max"
]

target = "RUL"

X_train = train_data[features]
y_train = train_data[target]

X_test = test_data[features]
y_test = test_data[target]

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

model = LinearRegression()

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("Linear Regression Results")
print("MAE:", mae)
print("RMSE:", rmse)
print("R2 Score:", r2)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.plot(
    y_test.values,
    label="Actual RUL"
)

plt.plot(
    y_pred,
    label="Predicted RUL"
)

plt.xlabel("Test Cycle")
plt.ylabel("Remaining Useful Life (cycles)")
plt.title("Actual vs Predicted RUL - Linear Regression")
plt.legend()
plt.show()
```

```python
xgb_model = xgb.XGBRegressor(
    n_estimators=200,
    max_depth=4,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="reg:squarederror",
    random_state=42
)

xgb_model.fit(X_train, y_train)

xgb_pred = xgb_model.predict(X_test)

print("XGBoost model trained successfully!")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

xgb_mae = mean_absolute_error(y_test, xgb_pred)
xgb_rmse = np.sqrt(mean_squared_error(y_test, xgb_pred))
xgb_r2 = r2_score(y_test, xgb_pred)

print("XGBoost Results")
print("MAE:", xgb_mae)
print("RMSE:", xgb_rmse)
print("R2 Score:", xgb_r2)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    y_test.values,
    label="Actual RUL"
)

plt.plot(
    xgb_pred,
    label="Predicted RUL"
)

plt.xlabel("Test Cycle")
plt.ylabel("Remaining Useful Life (cycles)")
plt.title("Actual vs Predicted RUL - XGBoost")
plt.legend()
plt.show()
```

```python
X_test_b18 = test_b18[features]
y_test_b18 = test_b18["RUL"]

b18_pred = xgb_model.predict(X_test_b18)

print("B0018 predictions generated successfully!")
print("Number of predictions:", len(b18_pred))
```

```python
b18_mae = mean_absolute_error(y_test_b18, b18_pred)
b18_rmse = np.sqrt(mean_squared_error(y_test_b18, b18_pred))
b18_r2 = r2_score(y_test_b18, b18_pred)

print("B0018 XGBoost Results")
print("MAE:", b18_mae)
print("RMSE:", b18_rmse)
print("R2 Score:", b18_r2)
```
